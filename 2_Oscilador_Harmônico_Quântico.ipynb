## **Potencial, Função de onda e Energia**

Nosso objetivo é novamente encontrar uma função de onda que satisfaça a Equação de Schrodinger, mas dessa vez para outro potencial:


$$
V(x) = \frac{1}{2} m \omega^2 x^2
$$

Analogamente ao caso clássico do oscilador, temos uma força restauradora que fará com que a partícula fique **confinada** (potencial parabólico). A equação de Schrodinger então se torna:

$$
-\frac{\hbar^2}{2m} \frac{d^2 \psi(x)}{dx^2} + \frac{1}{2} m \omega^2 x^2 \psi(x) = E \psi(x)
$$

Essa equação permite dois métodos para chegarmos em sua solução, sendo um deles através de manipulação de operadores, definindo os **operadores escada**, com a vantagem de que não precisamos resolver diretamente a equação de Schrodinger, enquanto o outro é puramente analítico, resolvendo em força bruta a equação diferencial.

Para facilitar, vamos dividir nossa teoria em seções diferentes para cada método abordado pelo livro.

---

### **Método Algébrico**

O método algébrico, consiste na definição de dois novos operadores utilizando os operadores momento e posição já conhecidos. Eles serão denominados **operadores escada**, por uma razão que logo ficará esclarecida. Vamos começar reescrevendo a equação de Schrodinger em função de dois operadores que conhecemos, **x** e **p**:

$$ \frac{1}{2m} \left[ p^2 + (m\omega x)^2 \right]\psi = E\psi $$

Nossa ideia é fatorar o Hamiltoniano:

$$ H = \frac{1}{2m} \left[ p^2 + (m\omega x)^2 \right] $$

No entanto, por serem **operadores** e não apenas números, eles não necessariamente comutam, ou seja, $xp$ é diferente de $px$. Assim, vamos definir dois novos operadores:

$$ a_{\pm} \equiv \frac{1}{\sqrt{2\hbar m\omega}} (\mp ip + m\omega x) $$

Dessa forma, ao multiplicarmos os novos operadores:

$$
\begin{aligned}
a_- a_+ &= \frac{1}{2\hbar m\omega} (ip + m\omega x)(-ip + m\omega x) \\
&= \frac{1}{2\hbar m\omega} [p^2 + (m\omega x)^2 - im\omega(xp - px)]
\end{aligned}
$$

Podemos visualizar o termo $(xp - px)$ que conforme explicado, não é necessariamente 0. Vamos utilizar a relação geral de comutação entre operadores A e B para simplificar nossa equação:

$$ [A,B] \equiv AB - BA $$

$$ a_- a_+ = \frac{1}{2\hbar m\omega} [p^2 + (m\omega x)^2] - \frac{i}{2\hbar}[x,p] $$

Assim, podemos usar uma função teste $f(x)$ para descobrimos a relação de comutação entre **p** e **x**:

$$ [x,p]f(x) = \left[ x \frac{\hbar}{i} \frac{d}{dx}(f) - \frac{\hbar}{i} \frac{d}{dx}(xf) \right] = \frac{\hbar}{i} \left( x \frac{df}{dx} - x \frac{df}{dx} - f \right) = i\hbar f(x) $$

$$ [x,p] = i\hbar $$

Esse resultado é conhecido como **relação de comutação canônica**! Dessa forma, nossa equação pode ser simplificada para:

$$ a_- a_+ = \frac{1}{\hbar\omega} H + \frac{1}{2} $$

Ou, isolando H:

$$ H = \hbar\omega \left( a_- a_+ - \frac{1}{2} \right) $$

Lembrando que a ordem dos operadores $a$ importa, ou seja:

$$ a_+ a_- = \frac{1}{\hbar\omega} H - \frac{1}{2} $$

Adicionalmente, a relação de comutação entre os operadores $a$ é dada por:

$$ [a_-, a_+] = 1 $$

Ou seja, a equação também pode ser escrita assim:

$$ H = \hbar\omega \left( a_+ a_- + \frac{1}{2} \right) $$

Podemos agrupar tudo da seguinte maneira:

$$ \hbar\omega \left( a_{\pm} a_{\mp} \pm \frac{1}{2} \right)\psi = E\psi $$

Observe que interessante, se $ψ$ satisfaz a equação de Schrodinger com energia $E$, então $a_+ψ$ satisfaz com energia $(E + ħω)$ e $a_-Ψ$ com energia $(E - ħω)$, ou seja:

$$
\begin{aligned}
H(a_+\psi) &= \hbar\omega \left( a_+ a_- + \frac{1}{2} \right) (a_+\psi) = \hbar\omega \left( a_+ a_- a_+ + \frac{1}{2} a_+ \right) \psi \\
&= \hbar\omega a_+ \left( a_- a_+ + \frac{1}{2} \right) \psi = a_+ \left[ \hbar\omega \left( a_+ a_- + 1 + \frac{1}{2} \right) \psi \right] \\
&= a_+ (H + \hbar\omega)\psi = a_+ (E + \hbar\omega)\psi = (E + \hbar\omega)(a_+\psi)
\end{aligned}
$$


$$
\begin{aligned}
H(a_-\psi) &= \hbar\omega \left( a_- a_+ - \frac{1}{2} \right) (a_-\psi) = \hbar\omega a_- \left( a_+ a_- - \frac{1}{2} \right) \psi \\
&= a_- \left[ \hbar\omega \left( a_- a_+ - 1 - \frac{1}{2} \right) \psi \right] = a_- (H - \hbar\omega)\psi = a_- (E - \hbar\omega)\psi \\
&= (E - \hbar\omega)(a_-\psi)
\end{aligned}
$$

Através dessas relações, vemos que podemos aplicar os operadores $a_+$ e $a_-$ sucessivas vezes para obter energias maiores e menores, justificando o nome "**operadores escada**" anteriormente comentado, de forma que $a_+$ é o **operador levantamento** e $a_-$ é o **operador de abaixamento**.

Agora que temos essas ferramentas, precisamos encontrar ao menos uma solução para aplicarmos repetidamente os operadores e obter todas as soluções que precisarmos! Nossa ideia irá se basear no teorema que diz que não podemos ter um valor de energia menor que o valor mínimo do potencial (*consulte o problema 2.2 do livro*), ou seja, **nesse caso nossa energia não pode ser menor que zero!** Podemos prosseguir construindo uma expressão para descobrir nosso degrau base de acordo com essa ideia:

$$a_- \psi_0 = 0$$

Assim:

$$ \frac{1}{\sqrt{2\hbar m\omega}} \left( \hbar \frac{d}{dx} + m\omega x \right)\psi_0 = 0 $$

Onde chegamos na seguinte equação diferencial:

$$ \frac{d\psi_0}{dx} = -\frac{m\omega}{\hbar} x \psi_0 $$

Ao resolve-la, temos:

$$ \psi_0(x) = A e^{-\frac{m\omega}{2\hbar}x^2} $$

Após normalizarmos e encontrar a constante $A$, temos a seguinte expressão para $ψ_0$:


$$ \psi_0(x) = \left( \frac{m\omega}{\pi\hbar} \right)^{1/4} e^{-\frac{m\omega}{2\hbar}x^2} $$

Por fim, para descobrirmos a energia basta inserirmos o resultado na Equação de Schrodinger na seguinte forma:

$$ \hbar\omega \left( a_{\pm} a_{\mp} \pm \frac{1}{2} \right)\psi = E\psi $$

Chegando em:

$$ E_0 = \frac{1}{2}\hbar\omega $$

Dessa forma, tendo o estado fundamental e a energia inicial, basta aplicarmos repetidamente os operadores de levantamento para descobrirmos os $n$ estados excitados e energias possíveis ($A_n$ é a constante de normalização de cada $ψ_0$):

$$ \psi_n(x) = A_n (a_+)^n \psi_0(x) $$

$$E_n = \left(n + \frac{1}{2}\right)\hbar\omega, \quad \text{para } n = 0, 1, 2, \dots$$

Se quisermos seguir pelo caminho totalmente álgebrico, o livro nos mostra como normalizar nossa função de onda de maneira algébrica. Como este método é opcional e mais trabalhoso (*Consulte o livro caso esteja interessado*), vamos focar nos resultados que ele nos traz:

$$ a_+ \psi_n = \sqrt{n+1} \psi_{n+1}, \quad a_- \psi_n = \sqrt{n} \psi_{n-1} $$

$$ \psi_n = \frac{1}{\sqrt{n!}} (a_+)^n \psi_0 $$

Finalizamos assim a resolução algébrica para o oscilador harmônico quântico! No próximo tópico, vamos mostrar a solução abordando um outro caminho (*note que não iremos resolver tudo novamente, apenas ressaltar os resultados importantes, se tiver interesse no passo a passo recomendo consultar o livro base desses notebook's*).

### **Método Analítico**

Para resolvermos diretamente a Equação de Schrodinger, Griffths sugere a introdução de uma variável adimensional:

\begin{aligned}
\xi &= \sqrt{\frac{m\omega}{\hbar}}\,x,
\end{aligned}

Reescreveendo a equação de Schrodinger:

$$
\begin{aligned}
\frac{d^2\psi}{d\xi^2} &= (\xi^2 - \lambda)\,\psi,
\quad
\lambda = \frac{2E}{\hbar\omega}\\[8pt]
\end{aligned}
$$

Ao resolvermos essa equação utilizando os métodos adequados, chegamos na mesma discretização da energia devido as condições de contorno:

$$E_n = \left(n + \frac{1}{2}\right)\hbar\omega, \quad \text{para } n = 0, 1, 2, \dots$$

E nossas funções de onda são dadas por:

$$ \psi_n(x) = \left( \frac{m\omega}{\pi\hbar} \right)^{1/4} \frac{1}{\sqrt{2^n n!}} H_n(\xi) e^{-\xi^2/2} $$

Onde $H_n(\xi)$ são os **polinômios de Hermite** (podem ser consultados em uma tabela ou obtidos através da **Fórmula de Rodrigues**) em função de nossa variável adimensional, de forma que ambas as soluções são equivalentes!

No primeiro caso, imaginamos os átomos ligados como massas conectadas por molas! O potencial dessa ligação pode ser aproximado (com grande semelhança em casos de pequenas vibrações) pelo oscilador harmônico quântico!

## **Explicação do código**
Para visualizar as propriedades calculadas anteriormente, foram construídos três gráficos:

- Azul para a função de onda do estado estacionário,
- Vermelho para a densidade de probabilidade daquele estado,
- Verde para a visualização dos níveis de energia conforme alteramos o estado e/ou frequência angular.

Dessa forma, através dos widgets "Estado (n)" e "Freq.(ω)", podemos visualizar nos gráficos os formatos de cada objeto.

---

## **Código**

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact
from scipy.special import eval_hermite
import math

# Para simplificar, usaremos m = 1 e hbar = 1.
def plot_oscilador_harmonico_v2(n, omega):
    n_max_plot = 15

# Construção do eixo x:
    L = np.sqrt(2 * (n_max_plot + 0.5) / omega) * 1.1
    x_min, x_max = -L, L
    x = np.linspace(x_min, x_max, 1000)

# Definição das variáveis
    xi = np.sqrt(omega) * x
    V_x = 0.5 * (omega**2) * (x**2)
    E_n = (n + 0.5) * omega

# Calculando psi(x) e |psi(x)|^2:
    N = (omega / np.pi)**0.25 / np.sqrt(2**n * math.factorial(n))
    psi = N * eval_hermite(n, xi) * np.exp(-xi**2 / 2)
    densidade_probabilidade = psi**2

# Criação dos três gráficos:
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

# Gráfico 1: Função de Onda
    ax1_v = ax1.twinx()
    ax1_v.plot(x, V_x, color='black', linestyle=':', alpha=0.3)
    ax1_v.set_yticks([]) # Remove números do eixo do potencial para não poluir

    ax1.plot(x, psi, color='blue', linewidth=2, label=fr'$\psi_{{{n}}}(x)$')
    ax1.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
    ax1.fill_between(x, psi, 0, color='blue', alpha=0.1)

    ax1.set_title(f'Função de Onda (n = {n})')
    ax1.set_xlabel('Posição ($x$)')
    ax1.set_ylabel(r'$\psi_n(x)$')

    amp = np.max(np.abs(psi))
    ax1.set_ylim(-amp * 1.3, amp * 1.3)
    ax1.set_xlim(x_min, x_max)

# Gráfico 2: Densidade de probabilidade
    ax2_v = ax2.twinx()
    ax2_v.plot(x, V_x, color='black', linestyle=':', alpha=0.3)
    ax2_v.set_yticks([])

    ax2.plot(x, densidade_probabilidade, color='red', linewidth=2)
    ax2.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
    ax2.fill_between(x, densidade_probabilidade, 0, color='red', alpha=0.2)

    ax2.set_title(f'Densidade de Prob. (n = {n})')
    ax2.set_xlabel('Posição ($x$)')
    ax2.set_ylabel(r'$|\psi_n(x)|^2$')

    prob_m = np.max(densidade_probabilidade)
    ax2.set_ylim(-0.05 * prob_m, prob_m * 1.2)
    ax2.set_xlim(x_min, x_max)

# Gráfico 3: Níveis de Energia
    ax3.plot(x, V_x, color='black', linewidth=2)

    for k in range(n_max_plot + 1):
        E_k = (k + 0.5) * omega
        x_turn = np.sqrt(2 * E_k / (omega**2))

        if k == n:
            ax3.hlines(y=E_k, xmin=-x_turn, xmax=x_turn, color='green', linewidth=3, zorder=5)
            ax3.text(x_turn * 1.1, E_k, f'$E_{k}$', color='green', va='center', fontweight='bold')
        else:
            ax3.hlines(y=E_k, xmin=-x_turn, xmax=x_turn, color='gray', linewidth=1, linestyle='--', alpha=0.4)

    ax3.set_title('Níveis de Energia e Potencial')
    ax3.set_xlabel('Posição ($x$)')
    ax3.set_ylabel('Energia')
    ax3.set_ylim(0, (n_max_plot + 1) * omega)
    ax3.set_xlim(x_min, x_max)

    plt.tight_layout()
    plt.show()

# Construção dos Widgets
interact(plot_oscilador_harmonico_v2,
         n=widgets.IntSlider(min=0, max=15, step=1, value=0, description='Estado (n):'),
         omega=widgets.FloatSlider(min=0.5, max=5.0, step=0.25, value=1.0, description=r'Freq.(ω):'))
```
