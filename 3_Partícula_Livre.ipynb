## **Potencial, Função de onda e Energia**

Dessa vez o potencial que nossa partícula está submetida é nulo! Ou seja,

$$V(x) = 0$$

Dessa forma, a equação de Schrodinger torna-se:

$$-\frac{ħ^2}{2m} \frac{d^2\psi(x)}{dx^2} = E\psi(x)$$

Podemos simplificar:

$$\frac{d^2\psi}{dx^2} = -k^2\psi, \quad \text{onde } k \equiv \frac{\sqrt{2mE}}{\hbar}$$

Repare que é a mesma equação para o poço quadrado infinito, então podemos propor:

$$\psi(x) = Ae^{ikx} + Be^{-ikx}$$

No entanto, **não temos condições de contorno**, ou seja, a partícula pode assumir **qualquer valor** de energia maior que zero. Assim, a função de onda dependente do tempo é:

$$\Psi(x,t) = Ae^{ik\left(x - \frac{\hbar k}{2m}t\right)} + Be^{-ik\left(x + \frac{\hbar k}{2m}t\right)}$$

Voltando para ondulatória clássica, sabemos que uma função qualquer dependente de $(x - vt)$ representa uma onda viajando para direita, enquanto relacionado a $(x + vt)$ representa uma onda viajando para esquerda. Assim, comparando os termos exponenciais:

$$\left(x \mp \frac{\hbar k}{2m}t\right) = (x ∓ vt) → v = \frac{\hbar k}{2m}$$

Assim, como os termos propagantes para direita e esquerda diferem apenas de um sinal, podemos agrupá-los:

$$\Psi_k(x,t) = Ae^{i\left(kx - \frac{\hbar k^2}{2m}t\right)}$$

Permitindo que $k$ seja negativo para não perdemos as ondas que se propagam para esquerda:

$$k \equiv \pm \frac{\sqrt{2mE}}{\hbar}, \quad \text{com } \begin{cases} k > 0 \Rightarrow \text{movendo-se para a direita}, \\ k < 0 \Rightarrow \text{movendo-se para a esquerda} \end{cases}$$

Dessa forma, os "estados estacionários" são as ondas que se propagam, de forma que o comprimento de onda e momento (relação de de Broglie) são dados por:

$$λ = \frac{2π}{|k|}$$

$$p = ħk$$

Reescrevendo a velocidade:

$$v_q = \frac{ħ|k|}{2m} = \sqrt{\frac{E}{2m}}$$

Se compararmos com a velocidade clássica de uma partícula com energia $E$:

$$v_c = \sqrt{\frac{2E}{m}} = 2v_q$$

Ou seja, a primeira vista nossa função de onda se move com **metade** da velocidade da partícula que ela representa. Mas, antes disso, temos um problema mais grave, **nossa função de onda não é normalizável**:

$$\int_{-\infty}^{+\infty} \Psi_k^* \Psi_k dx = |A|^2 \int_{-\infty}^{+\infty} dx = |A|^2 \infty$$

Ou seja, as soluções separáveis para partícula livre não representam estados físicos, **não temos partículas livres com energias definidas!** No entanto, ainda podemos escrever a solução geral para a equação de Schrodinger utilizando uma integral ao invés de uma soma discreta:

$$\Psi(x,t) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{+\infty} \phi(k) e^{i\left(kx - \frac{\hbar k^2}{2m}t\right)} dk$$

Diante disso, podemos normalizar essa solução escolhendo $\phi(k)$ adequado. Contudo, nossa função de onda carrega um **intervalo** de $k$, ou seja, um intervalo de energias e velocidades. Isso é chamado de **pacote de onda**. Podemos tomar o estado inicial da função de onda para determinarmos nosso $\phi(k)$:

$$\Psi(x,0) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{+\infty} \phi(k)e^{ikx} dk$$

Através da análise de Fourier, temos a resposta pelo teorema de Plancherel:

$$f(x) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{+\infty} F(k)e^{ikx} dk \Leftrightarrow F(k) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{+\infty} f(x)e^{-ikx} dx$$

Em que $F(k)$ é a transformada de Fourier de $f(x)$, enquanto $f(x)$ é a transformada inversa de Fourier de $F(k)$. Dessa maneira:

$$\phi(k) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{+\infty} \Psi(x,0)e^{-ikx} dx$$

Vamos retornar ao problema da velocidade da partícula agora. A ideia chave contida no livro nos diz que um pacote de ondas consiste em "ondulações" contidas em um "envelope", de forma que o correspondente da velocidade da partícula seria a velocidade do envelope (**velocidade do grupo**), não das ondulações individuais (**velocidade de fase**), ou seja, é preciso mostrar que **a velocidade de grupo é duas vezes a velocidade de fase**, se encaixando perfeitamente com a ideia da comparação com a velocidade clássica da partícula. Assim, o problema está em determinar a velocidade de grupo de um pacote de onda no seguinte formato geral (onde $ω$ é chamado de **relação de dispersão**):

$$\Psi(x,t) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{+\infty} \phi(k)e^{i(kx-\omega t)} dk$$

$$ω = \frac{ħk^2}{2m}$$

Ao resolver essa equação (consulte o livro para mais detalhes se tiver curiosidade), chegamos em:

$$v_{\text{grupo}} = \frac{d\omega}{dk}$$

$$v_{\text{fase}} = \frac{\omega}{k}$$

De forma que:

$$\frac{\omega}{k} = \frac{\hbar k}{2m}$$

$$\frac{d\omega}{dk} = \frac{\hbar k}{m}$$

Ou seja, confirmamos que a velocidade coincidente com a velocidade clássica da partícula é a do grupo, não a da fase dos estados estacionários como vimos anteriormente, portanto:

$$v_c = v_g = 2v_f$$

## **Explicação do código**
Para visualizar as propriedades calculadas anteriormente, foram construídos três gráficos:

- Azul para a função de onda no tempo,
- Vermelho para a densidade de probabilidade,
- Roxo para a visualização do espaço dos momentos de acordo com a transformada de Fourier para cada largura ($σ$) e momento ($k$).

Dessa forma, através dos widgets "Largura ($σ$)", "Tempo ($t$)" e "Momento ($k$)", podemos visualizar nos gráficos os formatos de cada objeto.

---

## **Código**

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def plot_particula_livre(t, k0, sigma):
    # Constantes (usando unidades atômicas onde hbar = 1 e massa m = 1)
    hbar = 1.0
    m = 1.0

    # Construção do eixo x:
    x_min, x_max = -20, 60
    x = np.linspace(x_min, x_max, 1000)

    # Velocidade de grupo clássica
    v_g = hbar * k0 / m

    # Fator de largura complexa que causa a dispersão com o tempo
    # sigma(t) = sigma^2 + i * hbar * t / (2m)
    sigma_t_complex = sigma**2 + 1j * hbar * t / (2 * m)

    # Fator de normalização inicial
    N = 1 / (2 * np.pi * sigma**2)**0.25

    # Função de onda psi(x,t) exata para um pacote gaussiano
    prefactor = N * (sigma / np.sqrt(sigma_t_complex))
    termo_exponencial = np.exp(- (x - v_g * t)**2 / (4 * sigma_t_complex))
    termo_fase = np.exp(1j * k0 * (x - hbar * k0 * t / (2 * m)))

    psi = prefactor * termo_exponencial * termo_fase

    # Partes extraídas para plotagem
    psi_real = np.real(psi)
    densidade_probabilidade = np.abs(psi)**2
    envoltoria = np.abs(psi)

    # Criação dos três gráficos:
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

    # Gráfico 1: Função de Onda e Envoltória
    ax1.plot(x, psi_real, color='blue', linewidth=1.5, label='Re($\psi$)')
    ax1.plot(x, envoltoria, color='black', linestyle='--', alpha=0.7, label='Envoltória')
    ax1.plot(x, -envoltoria, color='black', linestyle='--', alpha=0.7) # Envoltória inferior
    ax1.axhline(y=0, color='gray', linestyle='--', alpha=0.5)

    ax1.set_title(f'Função de Onda no Tempo (t = {t:.1f})')
    ax1.set_xlabel('Posição ($x$)')
    ax1.set_ylabel(r'$\psi(x,t)$')
    ax1.legend(loc='upper right')

    # Fixando o eixo Y baseado no pico mais alto possível (quando t=0 e sigma mínimo)
    amplitude_max = (1 / (2 * np.pi * 0.5**2)**0.25)
    ax1.set_ylim(-amplitude_max * 1.2, amplitude_max * 1.2)
    ax1.set_xlim(x_min, x_max)
    ax1.grid(True, alpha=0.3)

    # Gráfico 2: Densidade de probabilidade
    ax2.plot(x, densidade_probabilidade, color='red', linewidth=2)
    ax2.fill_between(x, densidade_probabilidade, 0, color='red', alpha=0.2)
    ax2.axhline(y=0, color='gray', linestyle='--', alpha=0.5)

    # Marcador do centro (posição esperada clássica)
    centro_x = v_g * t
    ax2.axvline(x=centro_x, color='green', linestyle=':', linewidth=2, label=f'Centro ($x_c = {centro_x:.1f}$)')

    ax2.set_title(r'Densidade de Probabilidade ($|\psi|^2$)')
    ax2.set_xlabel('Posição ($x$)')
    ax2.set_ylabel(r'$|\psi(x,t)|^2$')
    ax2.legend(loc='upper right')

    prob_max = 1 / np.sqrt(2 * np.pi * 0.5**2)
    ax2.set_ylim(-0.02, prob_max * 1.1)
    ax2.set_xlim(x_min, x_max)
    ax2.grid(True, alpha=0.3)

    # Gráfico 3: Espaço dos Momentos
    k_min, k_max = -2, 7
    k = np.linspace(k_min, k_max, 1000)

    # Transformada de Fourier (probabilidade no espaço dos momentos)
    phi_k_quadrado = np.sqrt(2 * sigma**2 / np.pi) * np.exp(-2 * sigma**2 * (k - k0)**2)

    ax3.plot(k, phi_k_quadrado, color='purple', linewidth=2)
    ax3.fill_between(k, phi_k_quadrado, 0, color='purple', alpha=0.2)
    ax3.axvline(x=k0, color='black', linestyle='--', label=f'$k_0 = {k0}$')

    ax3.set_title(r'Espaço dos Momentos $|\Phi(k)|^2$')
    ax3.set_xlabel('Número de Onda ($k \propto p$)')
    ax3.set_ylabel('Probabilidade')
    ax3.legend(loc='upper right')

    # Caixa de texto ilustrando o Princípio da Incerteza
    ax3.text(k_min + 0.5, np.max(phi_k_quadrado)*0.8,
             f'$\Delta x \propto \sigma = {sigma:.1f}$\n$\Delta p \propto 1/\sigma = {1/sigma:.2f}$',
             bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

    max_phi_y = np.sqrt(2 * 3.0**2 / np.pi) # Baseado no sigma máximo do slider
    ax3.set_ylim(0, max_phi_y * 1.1)
    ax3.set_xlim(k_min, k_max)
    ax3.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# Construção dos Widgets
interact(plot_particula_livre,
         t=widgets.FloatSlider(min=0.0, max=20.0, step=0.5, value=0.0, description='Tempo (t):'),
         k0=widgets.FloatSlider(min=0.0, max=5.0, step=0.5, value=2.0, description='Momento (k0):'),
         sigma=widgets.FloatSlider(min=0.5, max=3.0, step=0.5, value=1.0, description='Largura (σ):'))
```
