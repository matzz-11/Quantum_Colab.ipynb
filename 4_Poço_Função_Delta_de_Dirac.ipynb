### Função Delta de Dirac

A função delta de Dirac não é bem uma função, mas uma distribuição generalizada com as seguintes propriedades:

- Vale zero em todo local, exceto em x = 0
- Tem área total igual a 1.

$$
\int_{-\infty}^{\infty} δ(x) dx = 1
$$

- Quando aplicada em qualquer função contínua, geramos o valor da função em x = a.

$$
\int_{-\infty}^{\infty} f(x)δ(x-a)dx = f(a)
$$

Sua representação gráfica é um pico infinitamente estreito e alto, mas com área finita.

### Estados ligados e de espalhamento

Para facilitar o entendimento, Griffiths faz uma comparação com a mecânica clássica, mostrando o que acontece nas situações que temos um potencial $V(x)$ aumentando mais que a energia total da partícula $E$ nas duas direções ($+∞$ ou $-∞$) ou caso $V(x)$ aumente em apenas uma direção. Na primeira, sabemos que a partícula fica presa entre os pontos de retorno, sendo denomidado um **estado ligado**. Na segunda, a partícula pode "escapar" pelo lado que a energia $E$ aumenta mais que $V(x)$, sendo denominado **estado de espalhamento**.

Em nossa jornada na mecânica quântica vimos potenciais que admitem somente estados ligados (como o poço quadrado infinito) e somente estados de espalhamento (como a partícula livre), mas também teremos casos em que são admitdos ambos os casos.

$$
\begin{cases}
E < [V(-\infty) \quad \text{e} \quad V(+\infty)] \Rightarrow \quad \text{estado ligado,} \\
E > [V(-\infty) \quad \text{ou} \quad V(+\infty)] \Rightarrow \quad \text{estado de espalhamento.}
\end{cases}
$$

## **Potencial, função de onda e energia**

O potencial (o sinal negativo é devido ao estudo de um poço) é descrito da seguinte forma:

$$
V(x) = -αδ(x), α > 0
$$

Onde α é a "força" do nosso potencial. A equação de Schrodinger se torna:

$$
-\frac{\hbar^2}{2m} \frac{d^2\psi}{dx^2} - \alpha\delta(x)\psi = E\psi,
$$

Analisando a equação, podemos ter situações com $E < 0$ e $E > 0$, ou seja, **estados ligados e de espalhamento**. Seguindo nosso material de referência, vamos começar pelos estados ligados. Tomemos as regiões em $x > 0$ ou $x < 0$, onde $V(x) = 0$, de forma que a equação de Schrodinger é:

$$
\frac{d^2\psi}{dx^2} = -\frac{2mE}{\hbar^2}\psi = \kappa^2\psi
$$

Onde definimos:

$$
κ ≡ \frac{\sqrt{-2mE}}{ħ}
$$

Como tomamos $E$ negativo (estado ligado), temos κ como número real positivo. Já vimos a solução para essa equação diferencial para $x < 0$:

$$
\psi(x) = Be^{\kappa x}
$$

e $x > 0$:

$$
\psi(x) = Fe^{-\kappa x}
$$

Aplicando a primeira condição de contorno (continuidade de $ψ$), vemos que as constantes precisam ser iguais:

$$
\psi(x) = \begin{cases}
Be^{\kappa x}, & (x \le 0) \\
Be^{-\kappa x}, & (x \ge 0)
\end{cases}
$$

No entanto, essa função tem uma descontinuidade na derivada em $x = 0$. Utilizando $V(x) = -αδ(x)$ e aplicando a propriedade da função delta de Dirac no ponto $x = 0$ aplicada em outra função, teremos:

$$
\Delta\left(\frac{d\psi}{dx}\right) = -\frac{2m\alpha}{\hbar^2}\psi(0)
$$

Nesse caso:

$$
\Delta\left(\frac{d\psi}{dx}\right) = -2Bκ
$$

$$
ψ(0) = B
$$

Ou seja:

$$
κ = \frac{mα}{ħ^2}
$$

Temos então a **única** energia permitida, ou seja, apenas **um estado ligado**:

$$
E = -\frac{\hbar^2\kappa^2}{2m} = -\frac{m\alpha^2}{2\hbar^2}
$$

Normalizando a função de onda, encontramos:

$$
\psi(x) = \frac{\sqrt{m\alpha}}{\hbar} e^{-m\alpha|x|/\hbar^2}
$$

Prossigamos para os estados de espalhamento ($E > 0$):

$$
\frac{d^2\psi}{dx^2} = -\frac{2mE}{\hbar^2}\psi = -k^2\psi
$$

$$
k \equiv \frac{\sqrt{2mE}}{\hbar}
$$

Temos agora para $x < 0$:

$$
\psi(x) = Ae^{ikx} + Be^{-ikx}
$$

E para $x > 0$:

$$
\psi(x) = Fe^{ikx} + Ge^{-ikx}
$$

Novamente, pela continuidade temos:

$$
F + G = A + B
$$

Com:

$$
\Delta\left(\frac{d\psi}{dx}\right) = ik(F - G - A + B)
$$

Sabendo que $ψ(0) = (A + B)$:

$$
ik(F - G - A + B) = -\frac{2m\alpha}{\hbar^2}(A + B)
$$

Podemos compactar esse resultado da seguinte maneira:

$$
F - G = A(1 + 2i\beta) - B(1 - 2i\beta), \quad \text{em que} \quad \beta \equiv \frac{m\alpha}{\hbar^2 k}
$$

Como temos várias constantes e ao mesmo tempo estamos tratando de um estado não normalizável, vamos analisar o que essas constantes significam fisicamente. Repare que $e^{ikx}$ e $e^{-ikx}$ combinados com o fator temporal dão origem a ondas que se propagam para direita e esquerda, respectivamente.

Dessa forma, $A$ seria a amplitude de uma onda vinda da esquerda, $B$ a amplitudade de uma onda voltando para a esquerda, $F$ a amplitude de uma onda se movendo para direita e $G$ a amplitude de uma vinda da esquerda.

Para nosso experimento ter sentido físico, vamos escolher uma direção de onde essas ondas partem, nesse caso a esquerda. Assim, $G = 0$, pois não teremos ondas partindo da esquerda.

Podemos resumir da seguinte maneira:
- $A$ é a amplitude da onda incidente
- $B$ é a amplitude da onda refletida
- $C$ é a amplitude da onda transmitida

Ao resolver as equações para $B$ e $F$:

$$
B = \frac{i\beta}{1 - i\beta} A, \quad F = \frac{1}{1 - i\beta} A
$$

Como não podemos normalizar o estado e obter a probabilidade absoluta de encontrar a partícula, vamos trabalhar com a razão entre probabilidades, a probabilidade relativa. No caso de uma reflexão, temos:

$$
R \equiv \frac{|B|^2}{|A|^2} = \frac{\beta^2}{1 + \beta^2}
$$

E para uma transmissão:

$$
T \equiv \frac{|F|^2}{|A|^2} = \frac{1}{1 + \beta^2}
$$

Se somarmos as probabilidades, obtermos:

$$
R + T = \frac{\beta^2}{1 + \beta^2} + \frac{1}{1 + \beta^2} = 1
$$

Como de fato deveria ser, a soma das probabilidades resulta em 1, ou seja, a partícula precisa estar em algum lugar. Podemos por fim escrever as probabilidades $R$ e $T$ em função de $E$:

$$
R = \frac{1}{1 + (2\hbar^2 E / m\alpha^2)}, \quad T = \frac{1}{1 + (m\alpha^2 / 2\hbar^2 E)}
$$

## **Explicação do código**

Para visualizar as propriedades calculadas anteriormente, foram construídos três gráficos:

- Azul para a função de onda do estado estacionário,
- Vermelho para a densidade de probabilidade daquele estado,
- Roxo para o potencial no gráfico da energia,
- Verde para a visualização dos níveis de energia conforme alteramos o estado e/ou largura do poço quadrado infinito.

Dessa forma, através do widget "Força ($α$)" podemos visualizar nos gráficos os formatos de cada objeto.

---

## **Código**

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def plot_poco_delta(alpha):
    # Construção do eixo x
    x_min, x_max = -5, 5
    x = np.linspace(x_min, x_max, 1000)

    # Definição da função de onda e densidade de probabilidade
    psi = np.sqrt(alpha) * np.exp(-alpha * np.abs(x))
    densidade_probabilidade = psi**2

    # Criação dos três gráficos
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 5))

    # Gráfico 1: Função de Onda
    ax1.plot(x, psi, color='blue', linewidth=2)
    ax1.axvline(x=0, color='black', linestyle='--', alpha=0.3)
    ax1.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
    ax1.fill_between(x, psi, 0, color='blue', alpha=0.1)

    ax1.set_title(r'Função de Onda')
    ax1.set_xlabel('Posição ($x$)')
    ax1.set_ylabel(r'$\psi(x)$')

    amplitude_max = np.sqrt(5.0)
    ax1.set_ylim(-0.2, amplitude_max * 1.1)
    ax1.set_xlim(x_min, x_max)
    ax1.grid(True, alpha=0.3)

    # Gráfico 2: Densidade de probabilidade
    ax2.plot(x, densidade_probabilidade, color='red', linewidth=2)
    ax2.axvline(x=0, color='black', linestyle='--', alpha=0.3)
    ax2.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
    ax2.fill_between(x, densidade_probabilidade, 0, color='red', alpha=0.2)

    ax2.set_title(r'Densidade de Prob.')
    ax2.set_xlabel('Posição ($x$)')
    ax2.set_ylabel(r'$|\psi(x)|^2$')

    prob_max = 5.0
    ax2.set_ylim(-0.2, prob_max * 1.1)
    ax2.set_xlim(x_min, x_max)
    ax2.grid(True, alpha=0.3)

    # Gráfico 3: Nível de Energia (Corrigido)
    E = -(alpha**2) / 2  # Energia do estado ligado

    ax3.axhline(y=0, color='black', linewidth=1)
    ax3.axvline(x=0, color='gray', linestyle='--', alpha=0.3)

    # Limites do eixo Y fixos para ancorar os visuais
    E_min_grafico = -(5.0**2) / 2
    limite_inferior_y = E_min_grafico * 1.2
    ax3.annotate("", xy=(0, limite_inferior_y), xytext=(0, 0),
                 arrowprops=dict(arrowstyle="->", color='red', lw=2, linestyle='--'))
    ax3.text(0.2, limite_inferior_y * 0.8, r'$V(x) \to -\infty$', color='red', va='center')

    # Plotando o Nível de energia
    ax3.hlines(y=E, xmin=-2, xmax=2, color='green', linewidth=3, zorder=5)
    ax3.text(2.2, E, f'$E = {E:.2f}$', color='green', va='center', fontweight='bold', fontsize=12)

    ax3.set_title(r'Energia ($E = -\alpha^2 / 2$)')
    ax3.set_xlabel('Posição ($x$)')
    ax3.set_ylabel('Energia')

    ax3.set_ylim(limite_inferior_y, 2)
    ax3.set_xlim(x_min, x_max)
    ax3.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# Construção do Widget
interact(plot_poco_delta,
         alpha=widgets.FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description=r'Força (α):'))
```
