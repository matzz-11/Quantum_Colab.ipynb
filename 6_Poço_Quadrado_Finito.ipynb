## **Potencial, Função de onda e Energia**

Este caso é bem parecido com o poço infinito, com a diferença de que agora suas "paredes" não são mais infinitas! Como veremos adiante, isso fará com que **a função de onda possa "vazar" para fora do poço!**

Seu potencial é dado por:

$$
V(x) = \begin{cases}
-V_0 & \text{para } -a \leq x \leq a, \\
0, & \text{para } |x| > a,
\end{cases}
$$

Esse potencial ($V_0 > 0$) também admite estados ligados e de espalhamento, como vimos anteriormente no poço função Delta de Dirac. Vamos iniciar pelos estados ligados, dividindo regiões para facilitar.

Em $x < -a$ e $x > a$ o potencial é zero, resultando em:

$$
\frac{d^2\psi}{dx^2} = \kappa^2 \psi
$$

Com:

$$
\kappa \equiv \frac{\sqrt{-2mE}}{\hbar}
$$

Como vimos anteriormente, as soluções gerais para o primeiro e segundo caso são respectivamente:

$$
\psi(x) = A \exp(-\kappa x) + B \exp(\kappa x)
$$

$$
\psi(x) = F \exp(-\kappa x) + G \exp(\kappa x)
$$

No primeiro caso, a constante A deve ser zero pois sua exponencial diverge, enquanto analogamente G deve ser zero no segundo caso. Assim, ficamos com:

$$
\psi(x) = B e^{\kappa x}, \quad \text{para } x < -a
$$

$$
\psi(x) = F e^{-\kappa x}, \quad \text{para } x > a
$$

Vamos agora para a região $ -a < x < a$, onde temos o potencial como $-V_0$, de forma que:

$$
\frac{d^2\psi}{dx^2} = -l^2 \psi
$$

Com:

$$
l \equiv \frac{\sqrt{2m(E+V_0)}}{\hbar}
$$

A solução geral pode ser escrita como:

$$
\psi(x) = C \operatorname{sen}(lx) + D \cos(lx), \quad \text{para } -a < x < a
$$

Antes de impormos as condições de contorno, vamos analisar a paridade do potencial. Por ser par, podemos escolher soluções pares ou ímpares sem perda de generalidade. Além disso, nesse caso podemos impor as condições de contorno apenas em um lado, em $+a$ por exemplo. Dessa maneira, nossas soluções simplificadas se tornam:

$$
\psi(x) = \begin{cases}
F e^{-\kappa x} & \text{para } x > a, \\
D \cos(lx) & \text{para } 0 < x < a, \\
\psi(-x) & \text{para } x < 0.
\end{cases}
$$

Agora, impondo a continuidade de $ψ(x)$ e sua derivada temos respectivamente:

$$
F e^{-\kappa a} = D \cos(la)
$$

$$
\kappa F e^{-\kappa a} = l D \operatorname{sen}(la)
$$

Manipulando as duas equações, podemos escrever:

$$
\kappa = l \operatorname{tg}(la)
$$

Essa equação é a fórmula para energias permitidas, pois definimos $κ$ e $l$ como funções de $E$. Para simplificar, Griffiths propõe outras notações:

$$
z \equiv la \quad \text{e} \quad z_0 \equiv \frac{a}{\hbar} \sqrt{2mV_0}
$$

Que podemos usar para reescrever a equação como:

$$
\operatorname{tg} z = \sqrt{(z_0 / z)^2 - 1}
$$

Repare que se trata de uma equação transcendetal, a qual só pode ser resolvida graficamente ou numericamente (Não se preocupe, construi um código para que possamos visualizar as soluções!). Vamos analisar dois casos-limite interessantes:

- Poço largo e profundo:
Se temos $z_0$ muito grande, temos intersecções apenas um pouco abaixo de $z_n = nπ/2$, com $n$ ímpar. Temos então:

$$
E_n + V_0 \cong \frac{n^2 \pi^2 \hbar^2}{2m(2a)^2}
$$

Repare que já vimos essa fórmula de energia anteriormente, no caso do **poço quadrado infinito**. Ou seja, conforme $V_0 → ∞$, nosso poço quadrado finito tende ao poço quadrado infinito! **É importante notar que não importa quão grande $V_0$ fique, para qualquer valor finito teremos um número finito de estados ligados.**

- Poço raso e estreito:
Conforme $z_0$ fica menor, teremos um número cada vez menor de estados ligados, até que reste apenas um. **Não importa quão "fraco" nosso poço se torne, sempre haverá um estado ligado**.

Vamos agora trabalhar com os estados de espalhamento ($E > 0$). Em $x < -a$ ($V = 0$) temos:

$$
\psi(x) = A e^{ikx} + B e^{-ikx}
$$

Com:

$$
k \equiv \frac{\sqrt{2mE}}{\hbar}
$$

Para $x > a$ ($V = 0$) (Como estamos tratando de estados de espalhamento, vamos supor que não temos onda entrando pela direita, apenas pela esquerda como na solução acima):

$$
ψ(x) = F e^{ikx}
$$

Dentro do poço temos ($V(x) = -V_0$):

$$
\psi(x) = C \operatorname{sen}(lx) + D \cos(lx)
$$

Com:

$$
l \equiv \frac{\sqrt{2m(E+V_0)}}{\hbar}
$$

As constantes $A$, $B$ e $F$ serão denominadas amplitude incidente, amplitude refletida e amplitude transmitida, respectivamente. Aplicando as condições de contorno (continuidade de $ψ$ e sua derivada):

Em $x = -a$:
$$
A e^{-ika} + B e^{ika} = -C \operatorname{sen}(la) + D \cos(la)
$$

$$
ik[A e^{-ika} - B e^{ika}] = l[C \cos(la) + D \operatorname{sen}(la)]
$$

Em $x = a$:
$$
C \operatorname{sen}(la) + D \cos(la) = F e^{ika}
$$

$$
l[C \cos(la) - D \operatorname{sen}(la)] = ik F e^{ika}
$$

Podemos resolver para $B$ e $F$:

$$
B = i \frac{\operatorname{sen}(2la)}{2kl}(l^2 - k^2)F
$$

$$
F = \frac{e^{-2ika}A}{\cos(2la) - i\frac{(k^2+l^2)}{2kl}\operatorname{sen}(2la)}
$$

Analisando o coeficiente de transmissão ($T = \frac{|F|^2}{|A|^2}$):

$$
T^{-1} = 1 + \frac{V_0^2}{4E(E + V_0)} \operatorname{sen}^2\left( \frac{2a}{\hbar} \sqrt{2m(E + V_0)} \right)
$$

Vemos que $T = 1$ sempre que $\operatorname{sen} = 0$, que é dado por:

$$
\frac{2a}{\hbar} \sqrt{2m(E_n + V_0)} = n\pi
$$

Com $n$ sendo qualquer inteiro. Dessa forma, temos as energias para transmissões perfeitas dadas por:

$$
E_n + V_0 = \frac{n^2 \pi^2 \hbar^2}{2m(2a)^2}
$$

Que são exatamente as energias para o poço quadrado infinito.

## **Aplicações**

As aplicações do poço finito são basicamente as mesmas do poço infinito, mas **com maior precisão!** Isso pelo fato de que na realidade não existem barreiras infinitas de potencial, ou seja, na natureza **temos a probabilidade do tunelamento**, que não existe no modelo de poço infinito. Por conta disso, o poço quadrado finito acaba servindo melhor para as descrições de junção de semicondutores, band gaps, interações atômicas e ligações moleculares, que vimos anteriormente com uma aproximação menos precisa do poço infinito.

## **Código - Solução da Equação Transcendental por Gráfico**

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def resolucao_grafica_poco_finito(z0):
    # z0 define o limite do poço. z varia de um valor muito pequeno até z0.
    # Evitamos z=0 estrito para não causar divisão por zero.
    z = np.linspace(0.001, z0, 2000)

    # Lado esquerdo das equações: funções trigonométricas
    f_pares = np.tan(z)
    f_impares = -1.0 / np.tan(z) # -cot(z)

    # Lado direito da equação: a "curva limitante" do poço
    g = np.sqrt((z0 / z)**2 - 1)

    # Como tan(z) e -cot(z) são estritamente crescentes, uma queda brusca (derivada negativa) é uma assíntota.
    d_pares = np.diff(f_pares)
    f_pares[:-1][d_pares < 0] = np.nan

    d_impares = np.diff(f_impares)
    f_impares[:-1][d_impares < 0] = np.nan

    # Criação do gráfico
    plt.figure(figsize=(12, 7))

    # Plotando as funções
    plt.plot(z, f_pares, color='blue', linewidth=2.5, label=r'$\tan(z)$ (Estados Pares)')
    plt.plot(z, f_impares, color='red', linewidth=2.5, label=r'$-\cot(z)$ (Estados Ímpares)')
    plt.plot(z, g, color='green', linewidth=3, label=r'$\sqrt{(z_0/z)^2 - 1}$ (Limite do Poço)')

    # Marcações de Pi/2 no eixo X (onde ocorrem as assíntotas)
    max_pi = int(z0 / (np.pi / 2)) + 1
    xticks = []
    xticklabels = []
    for i in range(1, max_pi + 1):
        x_val = i * np.pi / 2
        plt.axvline(x=x_val, color='gray', linestyle='--', alpha=0.4)
        xticks.append(x_val)
        xticklabels.append(f'{i}$\pi/2$')

    # Adicionando a origem
    xticks = [0] + xticks
    xticklabels = ['0'] + xticklabels

    # Formatação do Gráfico
    plt.title(f'Resolução Gráfica - Interseções determinam $E_n$ ($z_0 = {z0:.1f}$)', fontsize=14)
    plt.xlabel('Variável adimensional ($z = la$)', fontsize=12)
    plt.ylabel('Amplitude das Funções', fontsize=12)
    plt.axhline(0, color='black', linewidth=1)

    plt.xticks(xticks, xticklabels)

    # Restringindo o eixo Y para focar nas interseções (ignorando o infinito)
    plt.ylim(0, 8)
    plt.xlim(0, z0 + 0.2)

    plt.legend(loc='upper right', fontsize=12)
    plt.grid(True, alpha=0.2)

    plt.tight_layout()
    plt.show()

# Widget
interact(resolucao_grafica_poco_finito,
         z0=widgets.FloatSlider(min=0.5, max=20.0, step=0.5, value=8.0, description='Força (z0):'))
```

## **Código - Poço**

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def plot_poco_finito(n, a, V0):
    # Construção do eixo x mais amplo (para ver o decaimento exponencial fora do poço)
    x_min, x_max = -3 * a, 3 * a
    x = np.linspace(x_min, x_max, 2000)

    # Equações Transcendentais do Griffiths
    # Usaremos unidades simplificadas (2m / hbar^2 = 1) para o cálculo.
    # z0 = a * sqrt(V0) e z = a * sqrt(E)
    z0 = a * np.sqrt(V0)
    z_vals = np.linspace(1e-5, z0 - 1e-5, 5000)

    # Funções do Griffiths cujas raízes nos dão os níveis de energia:
    # Estados Pares: z * sin(z) - sqrt(z0^2 - z^2) * cos(z) = 0
    # Estados Ímpares: z * cos(z) + sqrt(z0^2 - z^2) * sin(z) = 0
    g_even = z_vals * np.sin(z_vals) - np.sqrt(np.maximum(z0**2 - z_vals**2, 0)) * np.cos(z_vals)
    g_odd  = z_vals * np.cos(z_vals) + np.sqrt(np.maximum(z0**2 - z_vals**2, 0)) * np.sin(z_vals)

    roots_z = []
    # Procurando as raízes procurando por mudanças de sinal nas funções
    for g_vals in [g_even, g_odd]:
        sign_flips = np.where(np.diff(np.sign(g_vals)))[0]
        for idx in sign_flips:
            # Interpolação linear fina para achar a raiz com precisão
            z1, z2 = z_vals[idx], z_vals[idx+1]
            f1, f2 = g_vals[idx], g_vals[idx+1]
            root = z1 - f1 * (z2 - z1) / (f2 - f1)
            roots_z.append(root)

    roots_z = np.sort(roots_z)
    E_levels = (roots_z / a)**2  # Convertendo z de volta para Energia

    # Número de estados ligados
    num_estados = len(E_levels)
    if num_estados == 0:
        E_levels = np.array([V0 * 0.5]) # Prevenção de erro matemático
        num_estados = 1

    n_plot = min(n, num_estados) # Se n for maior que os estados possíveis, plotamos o maior
    E = E_levels[n_plot - 1]

    # Constantes de número de onda dentro (l) e decaimento fora (k) do poço
    l = np.sqrt(E)
    k = np.sqrt(V0 - E)

    # Definição da função de onda
    psi = np.zeros_like(x)
    regiao_esq = x < -a
    regiao_dir = x > a
    dentro_do_poco = (x >= -a) & (x <= a)

    # Estado ÍMPAR na contagem (n=1,3,5...) -> Função PAR no espaço (cossenos)
    if n_plot % 2 != 0:
        psi[dentro_do_poco] = np.cos(l * x[dentro_do_poco])
        psi[regiao_dir] = np.cos(l * a) * np.exp(-k * (x[regiao_dir] - a))
        psi[regiao_esq] = np.cos(l * a) * np.exp(k * (x[regiao_esq] + a))
    # Estado PAR na contagem (n=2,4,6...) -> Função ÍMPAR no espaço (senos)
    else:
        psi[dentro_do_poco] = np.sin(l * x[dentro_do_poco])
        psi[regiao_dir] = np.sin(l * a) * np.exp(-k * (x[regiao_dir] - a))
        psi[regiao_esq] = -np.sin(l * a) * np.exp(k * (x[regiao_esq] + a))

    # Normalização numérica da função inteira
    norm = np.sqrt(np.trapz(psi**2, x))
    psi = psi / norm
    densidade_probabilidade = psi**2

    # Criação dos gráficos
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))
    aviso = f' (n={n} não é ligado. Mostrando {n_plot})' if n > num_estados else ''

    # Gráfico 1: Função de Onda
    ax1.plot(x, psi, color='blue', linewidth=2)
    ax1.axvline(x=-a, color='black', linewidth=1.5, linestyle='--')
    ax1.axvline(x=a, color='black', linewidth=1.5, linestyle='--')
    ax1.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
    ax1.fill_between(x, psi, 0, where=dentro_do_poco, color='blue', alpha=0.1)
    ax1.fill_between(x, psi, 0, where=regiao_esq | regiao_dir, color='cyan', alpha=0.1)

    ax1.set_title(f'Função de Onda $\psi_{n_plot}(x)$ {aviso}')
    ax1.set_xlabel('Posição ($x$)')
    ax1.set_ylabel(r'$\psi_n(x)$')
    ax1.set_xlim(x_min, x_max)
    ax1.grid(True, alpha=0.3)

    # Gráfico 2: Densidade de Probabilidade
    ax2.plot(x, densidade_probabilidade, color='red', linewidth=2)
    ax2.axvline(x=-a, color='black', linewidth=1.5, linestyle='--')
    ax2.axvline(x=a, color='black', linewidth=1.5, linestyle='--')
    ax2.axhline(y=0, color='gray', linestyle='--', alpha=0.5)

    # Destacando o tunelamento em laranja
    ax2.fill_between(x, densidade_probabilidade, 0, where=dentro_do_poco, color='red', alpha=0.2)
    ax2.fill_between(x, densidade_probabilidade, 0, where=regiao_esq | regiao_dir, color='orange', alpha=0.4, label='Tunelamento')

    ax2.set_title(f'Densidade de Probabilidade $|\psi_{n_plot}(x)|^2$')
    ax2.set_xlabel('Posição ($x$)')
    ax2.set_xlim(x_min, x_max)
    ax2.legend()
    ax2.grid(True, alpha=0.3)

    # Gráfico 3: Níveis de Energia e Potencial V(x)
    potencial_x = [x_min, -a, -a, a, a, x_max]
    potencial_y = [V0, V0, 0, 0, V0, V0]
    ax3.plot(potencial_x, potencial_y, color='black', linewidth=2.5, label='V(x)')
    ax3.fill_between(potencial_x, potencial_y, V0, color='gray', alpha=0.1)

    # Níveis de energia
    for idx, Ek in enumerate(E_levels):
        k_n = idx + 1
        cor, lw, text_peso = ('green', 2.5, 'bold') if k_n == n_plot else ('gray', 1, 'normal')
        ax3.hlines(y=Ek, xmin=-3*a, xmax=3*a, color=cor, linewidth=lw, zorder=5)
        ax3.text(a * 1.1, Ek + V0*0.02, f'$E_{{{k_n}}}$', color=cor, fontweight=text_peso)

    ax3.set_title(f'Níveis de Energia ({num_estados} Estados Ligados)')
    ax3.set_xlabel('Posição ($x$)')
    ax3.set_ylabel('Energia ($E$)')
    ax3.set_ylim(-V0*0.05, V0*1.1)
    ax3.set_xlim(x_min, x_max)
    ax3.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# Widgets
interact(plot_poco_finito,
         n=widgets.IntSlider(min=1, max=10, step=1, value=1, description='Estado (n):'),
         a=widgets.FloatSlider(min=0.5, max=4.0, step=0.1, value=1.5, description='Semi-Larg(a):'),
         V0=widgets.FloatSlider(min=2.0, max=50.0, step=1.0, value=25.0, description='Profund(V0):'))
```
